```python
from pathlib import Path
import pandas as pd

path_set = Path.cwd()

header_files = sorted(list(path_set.glob("*header*.csv")))
if header_files:
  header_file = header_files[-1]
  df_header = pd.read_csv(header_file)
  print(f"Loaded Header File: {header_file.name}")
  display(df_header.head())
else:
  print("No header data files found in this folder.")

line_files = sorted(list(path_set.glob("*line*.csv")))
if line_files:
  line_file = line_files[-1]
  df_line = pd.read_csv(line_file)
  print(f"\nLoaded Line File: {line_file.name}")
  display(df_line.head())
else:
  print("No line data files found in this folder.")

premium_files = sorted(list(path_set.glob("*premium*.csv")))
if premium_files:
  premium_file = premium_files[-1]
  df_premium = pd.read_csv(premium_file)
  print(f"\nLoaded Premium File: {premium_file.name}")
  display(df_premium.head())
else:
  print("No premium data files found in this folder.")
```
